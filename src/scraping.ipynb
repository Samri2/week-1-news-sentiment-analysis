```python
# Core libraries
import pandas as pd
import numpy as np
import re
from datetime import datetime
from tqdm import tqdm

# The star of the show
from google_play_scraper import app, reviews, Sort

print("Libraries loaded successfully!")
```

```python
BANK_APPS = {
    "CBE": "com.combanketh.mobilebanking",
    "BOA": "com.boa.boaMobileBanking",
    "Dashen": "com.dashen.dashensuperapp"
}

REVIEWS_PER_BANK = 500
```

```python
def scrape_bank_reviews(bank_name, package_name, count=500):

    result, _ = reviews(
        package_name,
        lang="en",
        country="et",
        sort=Sort.NEWEST,
        count=count
    )

    cleaned_reviews = []

    for review in tqdm(result):

        cleaned_reviews.append({
            "review": review.get("content"),
            "rating": review.get("score"),
            "date": review.get("at").strftime("%Y-%m-%d"),
            "bank": bank_name,
            "source": "Google Play"
        })

    return cleaned_reviews
```

```python
all_reviews = []

for bank, package in BANK_APPS.items():

    print(f"Scraping {bank} reviews...")

    bank_reviews = scrape_bank_reviews(
        bank_name=bank,
        package_name=package,
        count=REVIEWS_PER_BANK
    )

    all_reviews.extend(bank_reviews)

print("Scraping completed.")
```

```python
df = pd.DataFrame(all_reviews)

print(df.shape)

df.head()
df.to_csv("../data/raw/fintech_reviews_raw.csv", index=False)

print("Dataset exported successfully.")
print(df.shape)
```
