# Data Preprocessing

This notebook cleans and prepares the scraped review dataset for analysis.

Cleaning Steps:
- Remove duplicates
- Handle missing values
- Normalize dates
- Standardize review text

```python
import pandas as pd
import numpy as np


df = pd.read_csv("../data/raw/fintech_reviews_raw.csv")
```

# to cheeck null values

```python
df.info()

df.isnull().sum()
```

to remove duplicate valuees

```python
duplicates = df.duplicated(subset=["review"]).sum()

print("Duplicates:", duplicates)

df.drop_duplicates(subset=["review"], inplace=True)

print("Duplicates after removal:", df.duplicated(subset=["review"]).sum())
```

```python
df.dropna(subset=["review", "rating"], inplace=True)
print("Null values after removal:", df.isnull().sum())
```

to normalaize text

```python
df["review"] = df["review"].astype(str).str.strip()
print("Text normalized")
```

to normalize dates

```python
df["date"] = pd.to_datetime(df["date"]).dt.strftime("%Y-%m-%d")
print("Date format normalized")
```

Final Dataset Check

```python
print(df.shape)

df.head()
```

```python
output_path = "../data/raw/fintech_reviews_clean.csv"

df.to_csv(output_path, index=False)

print("Dataset saved successfully.")
```
